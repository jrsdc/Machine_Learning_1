Estimate the parameters and find the best combination of hyperparameters (Code Reuse Session 2 (Week 5) snippet 24)

```python
# Import GridSearchCV for hyperparameter searching
from sklearn.model_selection import GridSearchCV
```

Based on the highest F1 score for class 0.0 at 0.58 the best chosen model will be KNN, in which case it will be tuned to find the best hyperparameter combnation (Code Reuse Session 2 (Week 5) snippet 25)

```python
# KNN is reinstantiated due to the fact that it was previously assigned to ROC,
# it will be make empty to begin the tuning process
knn = KNeighborsClassifier()
# Creating a dictionary for the hyperparameters to search through, n_neighbors controls how many points the
# algorithm consider during prediction
param_grid = {'n_neighbors': [1, 3, 5 , 7 , 9, 11, 13, 15]}
# instantiate grid seach to knn and take the arguements of knn for tuning, param_grid for testing and
# 5 fold cross validation for inner training
knn_gscv = GridSearchCV(knn, param_grid, cv = 5)
# Where GridSearchCV tries every number of n_neighbors to find the best combination
knn_gscv.fit(X_train,y_train)
# Displays the best combinaation
knn_gscv.best_params_
```

Use the optimised configurations to predict loan approval in the test data set (Code Reuse Session 2 (Week 5) snippet 26)

```python
# Used the tuned model with the hyperparameters for predictions on the test set
y_pred = knn_gscv.predict(X_test)
```
