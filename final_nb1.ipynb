This funcrtion detects outliers in any numeric variable based on the inter-quartile range test (Code Reuse Session 1 (Week 3) snippet 17)

```python
# Creating a function that can call the outliers of a specific numeric variable
def find_outliers_IQR(data):

  q1=data.quantile(0.25)

  q3=data.quantile(0.75)

  IQR=q3-q1

  outliers = data[((data<(q1-1.5*IQR))|(data> (q3+1.5*IQR)))]

  return outliers
```
